## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
data.info()
````

````python
# Import Libraries
import pandas as pd
import numpy as np

# Load Dataset
data = pd.read_csv('/content/drive/MyDrive/dataset.csv')

# -----------------------------
# 1. Remove Duplicate Records
# -----------------------------
data.drop_duplicates(inplace=True)

# -----------------------------
# 2. Replace Invalid Values
# -----------------------------
data.replace(['_', 'NA', '?', ''], np.nan, inplace=True)

# -----------------------------
# 3. Check Missing Values
# -----------------------------
print(data.isnull().sum())

# -----------------------------
# 4. Convert Data Types
# -----------------------------
numeric_columns = [
    'Age',
    'Annual_Income',
    'Monthly_Inhand_Salary',
    'Num_Bank_Accounts',
    'Num_Credit_Card',
    'Interest_Rate',
    'Num_of_Loan',
    'Delay_from_due_date',
    'Num_of_Delayed_Payment',
    'Changed_Credit_Limit',
    'Num_Credit_Inquiries',
    'Outstanding_Debt',
    'Credit_Utilization_Ratio',
    'Total_EMI_per_month',
    'Amount_invested_monthly',
    'Monthly_Balance'
]

for col in numeric_columns:
    data[col] = pd.to_numeric(data[col], errors='coerce')

# -----------------------------
# 5. Handle Missing Values
# -----------------------------
for col in numeric_columns:
   data[col].fillna(data[col].median(), inplace=True)

categorical_columns = data.select_dtypes(include='object').columns

for col in categorical_columns:
    data[col].fillna(data[col].mode()[0], inplace=True)

# -----------------------------
# 6. Remove Unnecessary Columns
# -----------------------------
data.drop(['ID', 'Customer_ID', 'Name', 'SSN'], axis=1, inplace=True)

# -----------------------------
# 7. Verify Dataset
# -----------------------------
print(data.info())
print(data.isnull().sum())
print(data.head())
````

*  The dataset initially contained duplicate records, which were removed to improve data quality.


*   Several columns had missing values and inconsistent entries (such as _, NA, and ?), which were successfully cleaned.

*  Some numerical columns were stored as object (string) data types and required conversion before analysis.

*   Identifier columns (ID, Customer_ID, Name, and SSN) were not useful for prediction and were removed.
*   Identifier columns (ID, Customer_ID, Name, and SSN) were not useful for prediction and were removed.
